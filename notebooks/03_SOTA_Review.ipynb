```python
# ============================================================
# 1.2 — Research Gap Analysis
# ============================================================

gap_analysis = """
╔══════════════════════════════════════════════════════════════════════════╗
║                    RESEARCH GAP ANALYSIS                               ║
╠══════════════════════════════════════════════════════════════════════════╣
║                                                                        ║
║  GAP 1: NO MULTIMODAL QUISHING DETECTION EXISTS                        ║
║  ──────────────────────────────────────────────                         ║
║  - Trad et al. (2025): visual only, no text understanding              ║
║  - Jain & Gupta (2022): text only, blind to QR images                  ║
║  - Bountakas et al. (2023): multimodal but for webpages, not QR        ║
║  → Q-Shield: FIRST to fuse QR visual + SMS semantic for quishing       ║
║                                                                        ║
║  GAP 2: PAYLOAD DECODING IS ASSUMED AND DANGEROUS                      ║
║  ────────────────────────────────────────────────                       ║
║  - Most methods (Nejati, Alnajim) require scanning/decoding the QR     ║
║  - Decoding = executing potential malicious redirect                    ║
║  - Only Trad et al. avoids decoding, but uses only ML (no DL)          ║
║  → Q-Shield: CNN-based detection WITHOUT decoding (zero-risk scan)     ║
║                                                                        ║
║  GAP 3: NO LIGHTWEIGHT/EDGE-DEPLOYABLE SOLUTION                        ║
║  ───────────────────────────────────────────────                        ║
║  - Bountakas uses ResNet + BERT (too heavy for mobile)                 ║
║  - DB-CBIL uses CNN + BiLSTM (not optimized for edge)                  ║
║  - No work targets mobile deployment for emerging markets              ║
║  → Q-Shield: MobileNetV2 + DistilBERT (designed for edge inference)    ║
║                                                                        ║
║  GAP 4: NO LOCAL CONTEXT FOR LATIN AMERICA                             ║
║  ──────────────────────────────────────────                             ║
║  - All existing work assumes English/global phishing patterns           ║
║  - Peruvian fintech ecosystem (Yape, Plin, BCP) has unique patterns    ║
║  - Social engineering uses local urgency cues and slang                 ║
║  → Q-Shield: FIRST to incorporate LatAm social engineering context     ║
║                                                                        ║
║  GAP 5: LIMITED EXPLAINABILITY                                         ║
║  ────────────────────────────────                                       ║
║  - Most methods are black-box classifiers                              ║
║  - Trad uses only feature importance (no visual explanations)          ║
║  → Q-Shield: SHAP (both branches) + Grad-CAM (visual attention maps)  ║
║                                                                        ║
╚══════════════════════════════════════════════════════════════════════════╝
"""

print(gap_analysis)
```

```python
# ============================================================
# 1.4 — PAPER FIGURE: Gap Visualization
# ============================================================
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

# Capability matrix: rows=methods, cols=capabilities
methods = ['Trad (2025)', 'Nejati (2025)', 'DB-CBIL (2024)',
           'Alnajim (2023)', 'Jain (2022)', 'Bountakas (2023)', 'Q-Shield']
capabilities = ['No QR\nDecoding', 'Visual\nAnalysis', 'Text\nSemantics',
                'Multimodal\nFusion', 'Edge\nDeployable', 'XAI', 'Local\nContext']

# 1 = has, 0 = lacks, 0.5 = partial
matrix = np.array([
    [1,   1,   0,   0,   0,   0.5, 0],    # Trad
    [0,   0.5, 0,   0,   0,   0.5, 0],    # Nejati
    [0.5, 1,   0.5, 1,   0,   0.5, 0],    # DB-CBIL
    [0,   0,   1,   0,   0,   0,   0],    # Alnajim
    [0.5, 0,   1,   0,   0,   0,   0],    # Jain
    [0.5, 1,   1,   1,   0,   0.5, 0],    # Bountakas
    [1,   1,   1,   1,   1,   1,   1],    # Q-Shield
])

fig, ax = plt.subplots(figsize=(10, 6))

# Color map: 0=red, 0.5=yellow, 1=green
from matplotlib.colors import ListedColormap
cmap = ListedColormap(['#e74c3c', '#f39c12', '#2ecc71'])

# Map continuous to discrete
discrete = np.zeros_like(matrix, dtype=int)
discrete[matrix == 0] = 0
discrete[matrix == 0.5] = 1
discrete[matrix == 1] = 2

im = ax.imshow(discrete, cmap=cmap, aspect='auto', vmin=0, vmax=2)

ax.set_xticks(range(len(capabilities)))
ax.set_xticklabels(capabilities, fontsize=9, ha='center')
ax.set_yticks(range(len(methods)))
ax.set_yticklabels(methods, fontsize=10)

# Add text annotations
labels = {0: '\u2717', 0.5: '~', 1: '\u2713'}  # cross, tilde, check
for i in range(len(methods)):
    for j in range(len(capabilities)):
        text = labels[matrix[i, j]]
        color = 'white' if matrix[i, j] != 0.5 else 'black'
        ax.text(j, i, text, ha='center', va='center', fontsize=14,
                fontweight='bold', color=color)

# Highlight Q-Shield row
ax.add_patch(plt.Rectangle((-0.5, 5.5), len(capabilities), 1,
             fill=False, edgecolor='blue', linewidth=3))

legend_elements = [
    mpatches.Patch(facecolor='#2ecc71', label='Supported'),
    mpatches.Patch(facecolor='#f39c12', label='Partial'),
    mpatches.Patch(facecolor='#e74c3c', label='Not Supported'),
]
ax.legend(handles=legend_elements, loc='upper left', bbox_to_anchor=(1.01, 1),
          fontsize=9, framealpha=0.9)

ax.set_title('Capability Comparison: Q-Shield vs State of the Art',
             fontweight='bold', fontsize=13, pad=15)

plt.tight_layout()
os.makedirs(os.path.join(BASE_DIR, 'figures'), exist_ok=True)
plt.savefig(os.path.join(BASE_DIR, 'figures', 'fig1_sota_gap_matrix.png'),
            dpi=300, bbox_inches='tight')
plt.savefig(os.path.join(BASE_DIR, 'figures', 'fig1_sota_gap_matrix.pdf'),
            bbox_inches='tight')
plt.show()
print('Saved: fig1_sota_gap_matrix.{png,pdf}')
```
